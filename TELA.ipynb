```python
import ipywidgets as widgets
from IPython.display import display, clear_output

# Lista global de produtos
produtos = ["Teclado", "Mouse", "Monitor"]

# Criando os elementos de Login
usuario_input = widgets.Text(description="Usuário:", placeholder="Digite o usuário")
senha_input = widgets.Password(description="Senha:", placeholder="Digite a senha")
botao_login = widgets.Button(description="Entrar", button_style="primary")
saida_login = widgets.Output()

# Container para abrigar a tela do sistema após login
sistema_container = widgets.Output()

def montar_sistema_estoque():
    with sistema_container:
        clear_output()

        # Título da Interface
        titulo = widgets.HTML("<h2>=== ESTOQUE FÁCIL ===</h2>")

        # Botão de Sair do Programa
        btn_sair = widgets.Button(description="Sair do Programa", button_style="danger", icon="sign-out")

        def realizar_logout(b):
            # Limpa o container do sistema
            with sistema_container:
                clear_output()
            # Limpa os campos de login
            usuario_input.value = ""
            senha_input.value = ""
            with saida_login:
                clear_output()
            # Mostra novamente a tela de login
            box_login.layout.display = 'flex'

        btn_sair.on_click(realizar_logout)

        # --- ABA 1: LISTAR PRODUTOS ---
        aba_listar = widgets.Output()
        def atualizar_lista():
            with aba_listar:
                clear_output()
                if not produtos:
                    print("Nenhum produto cadastrado.")
                else:
                    print("Produtos Cadastrados:")
                    for i, p in enumerate(produtos, 1):
                        print(f"{i} - {p}")

        # --- ABA 2: CADASTRAR PRODUTO ---
        txt_cadastrar = widgets.Text(description="Novo Produto:", placeholder="Nome do produto")
        btn_cadastrar = widgets.Button(description="Cadastrar", button_style="success")
        saida_cadastrar = widgets.Output()

        def cadastrar_produto(b):
            with saida_cadastrar:
                clear_output()
                nome = txt_cadastrar.value.strip().capitalize()
                if nome:
                    produtos.append(nome)
                    print(f"Produto '{nome}' cadastrado com sucesso!")
                    txt_cadastrar.value = ""
                    atualizar_lista()
                else:
                    print("Por favor, digite um nome válido!")

        btn_cadastrar.on_click(cadastrar_produto)
        box_cadastrar = widgets.VBox([txt_cadastrar, btn_cadastrar, saida_cadastrar])

        # --- ABA 3: ATUALIZAR PRODUTO ---
        txt_antigo = widgets.Text(description="Nome Antigo:", placeholder="Produto a ser alterado")
        txt_novo = widgets.Text(description="Novo Nome:", placeholder="Novo nome do produto")
        btn_atualizar = widgets.Button(description="Atualizar", button_style="warning")
        saida_atualizar = widgets.Output()

        def atualizar_produto(b):
            with saida_atualizar:
                clear_output()
                antigo = txt_antigo.value.strip().capitalize()
                novo = txt_novo.value.strip().capitalize()
                if antigo in produtos:
                    if novo:
                        posicao = produtos.index(antigo)
                        produtos[posicao] = novo
                        print(f"Produto '{antigo}' atualizado para '{novo}' com sucesso!")
                        txt_antigo.value = ""
                        txt_novo.value = ""
                        atualizar_lista()
                    else:
                        print("O novo nome não pode ser vazio!")
                else:
                    print("Produto não encontrado no estoque!")

        btn_atualizar.on_click(atualizar_produto)
        box_atualizar = widgets.VBox([txt_antigo, txt_novo, btn_atualizar, saida_atualizar])

        # --- ABA 4: EXCLUIR PRODUTO ---
        txt_excluir = widgets.Text(description="Produto:", placeholder="Produto a ser excluído")
        btn_excluir = widgets.Button(description="Excluir", button_style="danger")
        saida_excluir = widgets.Output()

        def excluir_produto(b):
            with saida_excluir:
                clear_output()
                alvo = txt_excluir.value.strip().capitalize()
                if alvo in produtos:
                    produtos.remove(alvo)
                    print(f"Produto '{alvo}' excluído com sucesso!")
                    txt_excluir.value = ""
                    atualizar_lista()
                else:
                    print("Produto não encontrado!")

        btn_excluir.on_click(excluir_produto)
        box_excluir = widgets.VBox([txt_excluir, btn_excluir, saida_excluir])

        # --- MONTAGEM DAS ABAS ---
        abas = widgets.Tab()
        abas.children = [aba_listar, box_cadastrar, box_atualizar, box_excluir]
        abas.set_title(0, "Listar")
        abas.set_title(1, "Cadastrar")
        abas.set_title(2, "Atualizar")
        abas.set_title(3, "Excluir")

        # Renderiza a lista inicial ao abrir o menu
        atualizar_lista()

        # Display de tudo incluindo o botão de Sair no topo
        display(widgets.HBox([titulo, btn_sair], layout=widgets.Layout(justify_content='space-between', align_items='center')), abas)

# Função que gerencia o clique de Login
def verificar_login(b):
    with saida_login:
        clear_output()
        if usuario_input.value == "ian" and senha_input.value == "1234":
            print("Login realizado com sucesso!")
            # Esconde o formulário de login
            box_login.layout.display = 'none'
            # Monta e ativa a tela do sistema dentro do container
            montar_sistema_estoque()
        else:
            print("Usuário ou senha inválidos!")

botao_login.on_click(verificar_login)

# Interface Visual de Login
titulo_login = widgets.HTML("<h3>Login - Estoque Fácil</h3>")
box_login = widgets.VBox([
    titulo_login,
    usuario_input,
    senha_input,
    botao_login,
    saida_login
], layout=widgets.Layout(align_items='flex-start', margin='10px'))

# Exibe a tela de login e prepara o container do sistema na tela
display(box_login)
display(sistema_container)
```
